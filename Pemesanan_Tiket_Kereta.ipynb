```python
# membuat variable
nama_pemesan = []
kode_kereta = []
nama_kereta = []
kode_kelas = []
nama_kelas = []
harga = []
```

```python
# Pengkondisian dan Perulangan
jumlah_pemesanan = int(input("Masukkan Jumlah Pemesanan : "))
for i in range (jumlah_pemesanan):
  print("Pesanan ke-", i+1)
  nama_pemesan.append(input('Masukkan nama pemesan : '))
  kode_kereta.append(input('Masukkan kode kereta |SRY|ARH| : '))
  kode_kelas.append(input('Masukkan kode kelas |E|B|P| : '))
  if kode_kereta[i]=="SRY" or kode_kereta[i]=="sry":
    nama_kereta.append('Serayu malam')
    if kode_kelas[i]=="E" or kode_kelas[i]=="e":
      nama_kelas.append('Ekonomi')
      harga.append(int(75000))
    elif kode_kelas[i]=="B" or kode_kelas[i]=="b":
      nama_kelas.append('Bisnis')
      harga.append(int(120000))
    elif kode_kelas[i]=="P" or kode_kelas[i]=="p":
      nama_kelas.append('Premiun')
      harga.append(int(150000))
    else:
      nama_kelas.append('Not Found')
      harga.append(int(0))
  elif kode_kereta[i]=="SRY" or kode_kereta[i]=="sry":
    nama_kereta.append('Argo Parahyangan')
    if kode_kelas[i]=="E" or kode_kelas[i]=="e":
      nama_kelas.append('Ekonomi')
      harga.append(int(180000))
    elif kode_kelas[i]=="B" or kode_kelas[i]=="b":
      nama_kelas.append('Bisnis')
      harga.append(int(220000))
    elif kode_kelas[i]=="P" or kode_kelas[i]=="p":
      nama_kelas.append('Premiun')
      harga.append(int(360000))
    else:
      nama_kelas.append('Not Found')
      harga.append(int(0))
  else:
    nama_kereta.append('Not Found')
    if kode_kelas[i]=="E" or kode_kelas[i]=="e":
      nama_kelas.append('Ekonomi')
      harga.append(int(0))
    elif kode_kelas[i]=="B" or kode_kelas[i]=="b":
      nama_kelas.append('Bisnis')
      harga.append(int(0))
    elif kode_kelas[i]=="P" or kode_kelas[i]=="p":
      nama_kelas.append('Premiun')
      harga.append(int(0))
    else:
      nama_kelas.append('Not Found')
      harga.append(int(0))
```

```python
# Perhitungan
total = sum(harga)
if total > 200000:
  diskon = 10/100*total
else:
  diskon = 0

bayar = total - diskon
```

```python
# Output / Keluaran
import pandas as pd
data = {
    'Nama':nama_pemesan,
    'kereta':nama_kereta,
    'kelas':nama_kelas,
    'Harga':harga
}
rekap_data = pd.DataFrame(data)
print('='*50)
print('\tPROGRAM PEMESANAN TIKET KERETA')
print('='*50)
print(rekap_data)
print('='*50)
print('Total Pesanan\t= ' ,total)
print('Diskon\t\t= ' ,diskon)
print('Total Bayar\t= ' ,total)
uang_bayar = int(input('Masukkan Uang\t: '))
kembalian = uang_bayar - bayar
print('Kembalian\t: ' ,kembalian)
print('='*50)
```
