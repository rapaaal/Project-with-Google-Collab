```python
import os

# Nama file tempat kita akan menyimpan daftar tugas
FILE_NAME = "todo_list.txt"

def load_tasks():
    """Membaca tugas dari file teks dan mengembalikannya sebagai list."""
    tasks = []
    # Mengecek apakah file sudah ada
    if os.path.exists(FILE_NAME):
        try:
            with open(FILE_NAME, 'r') as file:
                # Membaca setiap baris, menghapus spasi/newline berlebih di ujung string
                tasks = [line.strip() for line in file.readlines()]
        except Exception as e:
            print(f"Terjadi kesalahan saat membaca file: {e}")
    return tasks
```

```python
def save_tasks(tasks):
    """Menyimpan list tugas saat ini ke dalam file teks."""
    try:
        with open(FILE_NAME, 'w') as file:
            for task in tasks:
                file.write(task + "\n")
    except Exception as e:
        print(f"Terjadi kesalahan saat menyimpan file: {e}")
```

```python
def display_tasks(tasks):
    """Menampilkan daftar tugas dengan nomor urut ke layar."""
    print("\n" + "="*30)
    print("📋 DAFTAR TUGAS ANDA")
    print("="*30)

    if not tasks:
        print("Yeay! Tidak ada tugas saat ini. Anda bisa bersantai.")
    else:
        # Enumerasi digunakan untuk memberikan nomor urut mulai dari 1
        for index, task in enumerate(tasks, start=1):
            print(f"{index}. {task}")
    print("="*30)
```

```python
def add_task(tasks):
    """Meminta input pengguna dan menambahkannya ke list tugas."""
    new_task = input("\nMasukkan tugas baru: ").strip()

    if new_task:
        tasks.append(new_task)
        save_tasks(tasks) # Langsung simpan perubahan ke file
        print(f"✅ Tugas '{new_task}' berhasil ditambahkan!")
    else:
        print("❌ Tugas tidak boleh kosong.")
```

```python
def delete_task(tasks):
    """Menghapus tugas berdasarkan nomor urut yang diinputkan pengguna."""
    display_tasks(tasks)

    if not tasks:
        return # Keluar dari fungsi jika tidak ada tugas yang bisa dihapus

    try:
        # Meminta input nomor tugas yang ingin dihapus
        task_number = int(input("\nMasukkan nomor tugas yang ingin dihapus: "))

        # Validasi apakah nomor yang dimasukkan ada di dalam rentang list
        if 1 <= task_number <= len(tasks):
            removed_task = tasks.pop(task_number - 1)
            save_tasks(tasks) # Simpan perubahan ke file
            print(f"🗑️ Tugas '{removed_task}' berhasil dihapus!")
        else:
            print("❌ Nomor tugas tidak ditemukan.")
    except ValueError:
        # Menangani error jika pengguna memasukkan huruf/simbol alih-alih angka
        print("❌ Mohon masukkan angka yang valid.")
```

```python
def delete_task(tasks):
    """Menghapus tugas berdasarkan nomor urut yang diinputkan pengguna."""
    display_tasks(tasks)

    if not tasks:
        return # Keluar dari fungsi jika tidak ada tugas yang bisa dihapus

    try:
        # Meminta input nomor tugas yang ingin dihapus
        task_number = int(input("\nMasukkan nomor tugas yang ingin dihapus: "))

        # Validasi apakah nomor yang dimasukkan ada di dalam rentang list
        if 1 <= task_number <= len(tasks):
            removed_task = tasks.pop(task_number - 1)
            save_tasks(tasks) # Simpan perubahan ke file
            print(f"🗑️ Tugas '{removed_task}' berhasil dihapus!")
        else:
            print("❌ Nomor tugas tidak ditemukan.")
    except ValueError:
        # Menangani error jika pengguna memasukkan huruf/simbol alih-alih angka
        print("❌ Mohon masukkan angka yang valid.")

def main():
    """Fungsi utama yang menjalankan loop menu aplikasi."""
    # Muat tugas dari file saat program pertama kali dijalankan
    tasks = load_tasks()

    while True:
        # Menampilkan menu utama
        print("\n--- MENU TO-DO LIST ---")
        print("1. Lihat Daftar Tugas")
        print("2. Tambah Tugas")
        print("3. Hapus Tugas")
        print("4. Keluar")

        choice = input("Pilih menu (1/2/3/4): ").strip()

        # Logika percabangan berdasarkan pilihan pengguna
        if choice == '1':
            display_tasks(tasks)
        elif choice == '2':
            add_task(tasks)
        elif choice == '3':
            delete_task(tasks)
        elif choice == '4':
            print("\n👋 Terima kasih telah menggunakan aplikasi To-Do List. Sampai jumpa!")
            break # Keluar dari loop / Menghentikan program
        else:
            print("❌ Pilihan tidak valid. Silakan pilih 1, 2, 3, atau 4.")

if __name__ == "__main__":
    main()
```
